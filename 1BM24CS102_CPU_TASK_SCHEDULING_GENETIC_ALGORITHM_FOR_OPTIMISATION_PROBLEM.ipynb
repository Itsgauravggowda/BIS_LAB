```python
import random

tasks = [4, 6, 8, 3, 5, 7, 2, 9]
num_cpus = 3
population_size = 10
generations = 100
mutation_rate = 0.1


def create_individual():
    return [random.randint(0, num_cpus - 1) for _ in tasks]


def create_population():
    return [create_individual() for _ in range(population_size)]


def fitness(individual):
    cpu_load = [0] * num_cpus

    for i in range(len(tasks)):
        cpu_load[individual[i]] += tasks[i]

    return 1 / max(cpu_load)


def selection(population):
    selected = random.sample(population, 3)
    return max(selected, key=fitness)


def crossover(parent1, parent2):
    point = random.randint(1, len(tasks) - 1)
    return parent1[:point] + parent2[point:]


def mutation(individual):
    for i in range(len(individual)):
        if random.random() < mutation_rate:
            individual[i] = random.randint(0, num_cpus - 1)

    return individual


def evaluate_population(population):
    return sorted(population, key=fitness, reverse=True)


def replace_population(population, new_population):
    combined = population + new_population
    combined = evaluate_population(combined)
    return combined[:population_size]


def genetic_algorithm():
    population = create_population()

    for generation in range(generations):
        population = evaluate_population(population)
        new_population = []

        while len(new_population) < population_size:
            parent1 = selection(population)
            parent2 = selection(population)

            child = crossover(parent1, parent2)
            child = mutation(child)

            new_population.append(child)

        population = replace_population(population, new_population)

    return max(population, key=fitness)


best_schedule = genetic_algorithm()
cpu_load = [0] * num_cpus

for i, cpu in enumerate(best_schedule):
    cpu_load[cpu] += tasks[i]
    print(f"Task {i + 1} -> CPU {cpu + 1}")

for i in range(num_cpus):
    print(f"CPU {i + 1}: {cpu_load[i]}")

print("Minimum Maximum Execution Time:", max(cpu_load))
print("Fitness Value:", fitness(best_schedule))
```
